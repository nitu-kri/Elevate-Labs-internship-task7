```python
rows = cursor.execute("select product_name, sum(1* amount) as total_sales from sales group by product_name").fetchall()
print(rows)
```

```python
rows = cursor.execute("select count(distinct product_id) as order_volume from sales").fetchall()
print(rows)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
```

```python
query = "SELECT product_name, sum(amount) as total_sales from sales group by product_name"
df_sales = pd.read_sql(query, connection)
plt.figure(figsize=(8,5))
sns.barplot(x="product_name", y="total_sales", data=df_sales, palette="coolwarm")
plt.xticks(rotation=45)
plt.title("total sales by product name")
plt.show()
```
